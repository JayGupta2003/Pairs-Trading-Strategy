```python
import sys
import os
import pandas as pd
```

```python
sys.path.append(os.path.abspath(os.path.join('..', 'src')))
from data_loader import get_sector_files,extract_tickers_from_csv, download_sector_price_data # type: ignore
from analytics import find_cointegrated_pairs # type: ignore
from screener import run_pdf_screener # type: ignore
```

```python
sector_files = get_sector_files()
all_valid_pairs = []
print(f"Found {len(sector_files)} sector files to process.")
```

```python
for file_path in sector_files:
    sector_name = os.path.basename(file_path).replace('.csv', '')
    print(f"--- Processing Sector: {sector_name} ---")

    tickers = extract_tickers_from_csv(file_path)
    print(f"  Found {len(tickers)} tickers.")
    
    if len(tickers) < 2:
        continue

    prices = download_sector_price_data(tickers, '2020-01-01', '2025-12-04')
    if prices.empty:
        print("No price data found.")
        continue

    prices.dropna(inplace=True)

    _, pairs = find_cointegrated_pairs(prices)
    print(f"  ✅ Found {len(pairs)} pairs in {sector_name}")
    
    for p in pairs:
        all_valid_pairs.append({
            'Stock A': p[0], 
            'Stock B': p[1], 
            'P-Value': p[2],
            'Sector': sector_name
        })

print(f"\n✅ Scan Complete. Total Pairs Found: {len(all_valid_pairs)}")
```

```python
results_df = pd.DataFrame(all_valid_pairs).sort_values('P-Value')
results_df.to_csv('../data/best_pairs.csv', index=False)
```
